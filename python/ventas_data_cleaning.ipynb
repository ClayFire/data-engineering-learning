```python
# Pandas es una librería de Python esencial para manipular, limpiar y analizar datos de forma fácil.
import pandas as pd
```

```python
# Creamos un dataframe con los siguientes valores y luego definimos el dataframe como df.
data = {
    "cliente": ["Ana", "Pedro", "Ana", "Maria", "Pedro"],
    "producto": ["Laptop", "Mouse", "Teclado", "Laptop", "Monitor"],
    "precio": [900000, 20000, 50000, 850000, 200000],
    "region": ["Santiago", "Valparaiso", "Santiago", "Concepcion", "Valparaiso"]
}

df = pd.DataFrame(data)

df
```

```python
# La función df.describe() nos permite conocer los valores estadísticos de las columnas que sean de valores cuantitativos.
df.describe()
```

```python
# Definimos una variable llamada ventas_region, en donde calculará a través del group by las sumas de las ventas por región.
ventas_region = df.groupby("region")["precio"].sum()

ventas_region
```

```python
# A través de la mención al dataframe por su abreviación df podemos solicitar el conteo de las veces que aparece cada producto.
df["producto"].value_counts()
```

```python
# Para guardar nuestro dataset, se puede ejecutar el siguiente comando para crearlo como archivo ".csv".
df.to_csv("ventas_limpias.csv", index=False)
```
